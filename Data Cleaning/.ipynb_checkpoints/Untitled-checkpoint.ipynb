```python
import pandas as pd
```

## 2012 Data Cleaning

```python
pell_2012=pd.read_csv("Pell Grant/Pell_2012.csv")
race_2012=pd.read_csv("Race/2012_Race.csv")
tuition_2012_2016=pd.read_csv("Tuition and Fees/Tuition_2012-2016.csv")
```

```python
columns_to_keep_12=['unitid', 'institution name','IC2015_AY.Published out-of-state tuition and fees 2012-13','IC2015_AY.Published in-state tuition and fees 2012-13', 'IC2015_AY.Published in-district tuition and fees 2012-13']
tuition_2012=tuition_2012_2016[columns_to_keep_12].copy()
tuition_2012.head()

merge_2012_1= pd.merge(pell_2012, race_2012, on=['unitid', 'year','institution name'], how='inner')
merge_2012_1.head()

merge_2012_final=pd.merge(merge_2012_1, tuition_2012, on=['unitid','institution name'], how='inner')
merge_2012_final
```
