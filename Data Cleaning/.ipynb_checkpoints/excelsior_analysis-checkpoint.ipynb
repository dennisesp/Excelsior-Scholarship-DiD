# Excelsior Scholarship: Replication and QuantCrit Extension

**Part A** replicates Howard (2026) for public institutions: counts (not logs), institution and year fixed effects, standard errors clustered by institution, 2016 as the event-study reference year, treatment in 2017 (and 2018 as a robustness check), and covariates of logged tuition and lagged county unemployment.

**Part B** is the QuantCrit-informed extension: every IPEDS category, three outcome measures (counts, composition, population-adjusted), group-specific pre-trend tests, institutional context (CUNY vs. SUNY), and a sensitivity check showing how sample decisions change the estimate.

Before running: at the end of your data-cleaning notebook, run `panel.to_csv("panel_clean.csv", index=False)`.

```python
import numpy as np
import pandas as pd
import statsmodels.formula.api as smf
import matplotlib.pyplot as plt

pd.set_option("display.float_format", "{:.3f}".format)

panel = pd.read_csv("panel_clean.csv", dtype={"fips": str})

# short names (ignored if you already renamed them)
panel = panel.rename(columns={
    "Grand total": "ftft",
    "Number of full-time first-time undergraduates awarded Pell grants": "pell",
    "Published in-state tuition and fees": "tuition",
    "Black or African American total": "black",
    "Hispanic total": "hispanic",
    "White total": "white",
    "Asian total": "asian",
    "American Indian or Alaska Native total": "aian",
    "Native Hawaiian or Other Pacific Islander total": "nhpi",
    "Two or more races total": "two_plus",
    "Race/ethnicity unknown total": "unknown",
    "Nonresident alien total": "nonresident",
    "U.S. Nonresident total": "nonresident",
})

panel["ny"] = (panel["state"] == "NY").astype(int)
panel["log_tuition"] = np.log(panel["tuition"])
panel["cuny"] = panel["institution name"].str.contains("CUNY", na=False).astype(int)

print(panel.groupby(["fall_year", "state"])["unitid"].nunique().unstack())
print(panel[["ftft", "pell", "tuition", "unemp_lag"]].isna().sum())
```

## Helper functions

These mirror Howard's Stata workflow: `didregress` for difference-in-differences, an event study with 2016 omitted, and `estat ptrends` (a Wald test of whether New York and Pennsylvania had different linear slopes before treatment).

```python
COVARS = "log_tuition + unemp_lag"
EVENT_YEARS = [y for y in range(2012, 2020) if y != 2016]

for y in EVENT_YEARS:
    panel[f"ny_{y}"] = panel["ny"] * (panel["fall_year"] == y).astype(int)
ES_TERMS = " + ".join(f"ny_{y}" for y in EVENT_YEARS)


def fit(formula, data):
    data = data.dropna(subset=[c for c in data.columns if c in formula])
    return smf.ols(formula, data=data).fit(cov_type="cluster", cov_kwds={"groups": data["unitid"]})


def did(outcome, data=None, treat_year=2017, covariates=False):
    """Difference-in-differences (Howard Tables 10-11)."""
    d = (panel if data is None else data).copy()
    d["treat_post"] = d["ny"] * (d["fall_year"] >= treat_year).astype(int)
    f = f"{outcome} ~ treat_post + C(unitid) + C(fall_year)" + (f" + {COVARS}" if covariates else "")
    m = fit(f, d)
    return {"coef": m.params["treat_post"], "se": m.bse["treat_post"],
            "t": m.tvalues["treat_post"], "p": m.pvalues["treat_post"], "n": int(m.nobs)}


def pretrend_test(outcome, data=None, treat_year=2017):
    """Equivalent of Stata's estat ptrends: Wald test of different NY vs. PA linear slopes before treatment (Howard Tables 8-9)."""
    d = (panel if data is None else data)
    d = d[d["fall_year"] < treat_year].copy()
    d["ny_trend"] = d["ny"] * (d["fall_year"] - d["fall_year"].min())
    m = fit(f"{outcome} ~ ny_trend + C(unitid) + C(fall_year)", d)
    return {"slope_diff": m.params["ny_trend"], "F": m.tvalues["ny_trend"] ** 2, "p": m.pvalues["ny_trend"]}


def event_study(outcome, data=None, covariates=False):
    """Event study with 2016 as reference (Howard Table 12)."""
    d = (panel if data is None else data)
    f = f"{outcome} ~ {ES_TERMS} + C(unitid) + C(fall_year)" + (f" + {COVARS}" if covariates else "")
    m = fit(f, d)
    ci = m.conf_int()
    out = pd.DataFrame({"coef": m.params[[f"ny_{y}" for y in EVENT_YEARS]],
                        "se": m.bse[[f"ny_{y}" for y in EVENT_YEARS]]})
    out["ci_low"], out["ci_high"] = ci.loc[out.index, 0], ci.loc[out.index, 1]
    out["p"] = m.pvalues[out.index]
    out.index = EVENT_YEARS
    pre = m.f_test(", ".join(f"ny_{y} = 0" for y in [2012, 2013, 2014, 2015]))
    post = m.f_test(", ".join(f"ny_{y} = 0" for y in [2017, 2018, 2019]))
    return out, {"pre_F": float(pre.fvalue), "pre_p": float(pre.pvalue),
                 "post_F": float(post.fvalue), "post_p": float(post.pvalue), "n": int(m.nobs)}


def plot_event_study(es, title, ax=None):
    ax = ax or plt.subplots(figsize=(7, 4))[1]
    full = pd.concat([es, pd.DataFrame({"coef": [0], "ci_low": [0], "ci_high": [0]}, index=[2016])]).sort_index()
    ax.errorbar(full.index, full["coef"], yerr=[full["coef"] - full["ci_low"], full["ci_high"] - full["coef"]],
                fmt="o", capsize=3, color="black")
    ax.axhline(0, color="gray", lw=0.8)
    ax.axvline(2016.5, color="gray", ls="--", lw=0.8)
    ax.set_title(title)
    ax.set_xlabel("Fall of entry")
    ax.set_ylabel("Difference vs. 2016 (students)")
    return ax
```

# Part A: Replication of Howard (2026)

## A1. Descriptive statistics (Howard Table 2 and Table 4 equivalent)

```python
base = panel[panel["fall_year"] == 2016]
desc = base.groupby("state").agg(
    institutions=("unitid", "nunique"),
    mean_ftft=("ftft", "mean"),
    mean_pell=("pell", "mean"),
    mean_tuition=("tuition", "mean"),
    mean_unemp=("unemp_lag", "mean"),
).T
desc
```

## A2. Parallel trends: aggregate outcomes (Howard p. 61-62)

```python
rows = []
for y in ["ftft", "pell"]:
    for t in [2017, 2018]:
        r = pretrend_test(y, treat_year=t)
        rows.append({"outcome": y, "treatment_year": t, **r})
pd.DataFrame(rows)
```

## A3. Difference-in-differences (Howard Tables 10-11, public institutions)

Howard's public-institution estimates with treatment in 2017: FTFT 45.17 (SE 39.42) without covariates, 22.07 (49.45) with; Pell FTFT 44.18 (22.95) and 18.60 (25.70). Your sample differs (Penn State combined), so expect similar direction and size, not identical numbers.

```python
rows = []
for y in ["ftft", "pell"]:
    for t in [2017, 2018]:
        for cov in [False, True]:
            rows.append({"outcome": y, "treatment_year": t, "covariates": cov,
                         **did(y, treat_year=t, covariates=cov)})
did_table = pd.DataFrame(rows)
did_table
```

## A4. Event study (Howard Table 12)

Howard's public FTFT estimates: 2017 = 68.35, 2018 = 83.29, 2019 = 79.21; pre-trends F = 0.61, post F = 4.02.

```python
es_ftft, tests_ftft = event_study("ftft")
es_pell, tests_pell = event_study("pell")
print("FTFT:", {k: round(v, 3) for k, v in tests_ftft.items()})
print("Pell:", {k: round(v, 3) for k, v in tests_pell.items()})
es_ftft
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4), sharex=True)
plot_event_study(es_ftft, "First-time, full-time enrollment", axes[0])
plot_event_study(es_pell, "Pell recipients (FTFT)", axes[1])
plt.tight_layout()
plt.savefig("fig2_event_study.png", dpi=200)
```

## A5. Race-specific pre-trend tests (Howard Table 8)

This is where Howard stopped. For public institutions he found Hispanic (p = .01) and White (p = .02) pre-trends diverged, so he abandoned race-specific estimates.

```python
rows = []
for g in ["black", "hispanic", "white"]:
    for t in [2017, 2018]:
        rows.append({"group": g, "treatment_year": t, **pretrend_test(g, treat_year=t)})
pd.DataFrame(rows)
```

# Part B: QuantCrit-informed extension

## B1. What the categories contain (*categories are neither natural nor given*)

Before analyzing any group, document every IPEDS category, including the ones conventional analyses leave out. Watch the Unknown and Two or More Races shares: if either shifts around 2017, it can produce apparent changes in other groups.

```python
ALL_GROUPS = [g for g in ["black", "hispanic", "white", "asian", "aian", "nhpi", "two_plus", "unknown", "nonresident"]
              if g in panel.columns]

comp = panel.groupby(["state", "fall_year"])[ALL_GROUPS + ["ftft"]].sum()
comp_share = comp[ALL_GROUPS].div(comp["ftft"], axis=0) * 100
comp_share.round(1)
```

```python
# check whether the categories add up to the total (gaps signal reporting problems)
check = panel[ALL_GROUPS].sum(axis=1) - panel["ftft"]
print("Rows where categories don't sum to total:", (check != 0).sum())
```

## B2. Three outcome measures (*numbers are not neutral*)

1. **Counts**: how many students enrolled (Howard's measure)
2. **Composition**: each group's share of an institution's first-time enrollment
3. **Population-adjusted**: enrollment relative to the state's 12th-grade population of that group (Howard p. 66-69, CCD data advanced one year)

For measure 3 you need CCD 12th-grade counts by state, race, and year saved as `ccd_12th.csv` with columns `state, fall_year, black, hispanic, white, asian` (12th graders in fall 2015 → `fall_year` 2016). If you don't have it yet, the cell skips it.

```python
for g in ["black", "hispanic", "white", "asian"]:
    panel[f"{g}_share"] = panel[g] / panel["ftft"] * 100

try:
    ccd = pd.read_csv("ccd_12th.csv")
    ccd = ccd.rename(columns={g: f"{g}_12th" for g in ["black", "hispanic", "white", "asian"]})
    panel = panel.merge(ccd, on=["state", "fall_year"], how="left")
    for g in ["black", "hispanic", "white", "asian"]:
        panel[f"{g}_per1k"] = panel[g] / panel[f"{g}_12th"] * 1000   # enrollees per 1,000 12th graders
    HAVE_CCD = True
except FileNotFoundError:
    HAVE_CCD = False
    print("No CCD file yet: population-adjusted measure skipped.")
```

## Figure 1: trends by race, counts vs. composition

```python
groups = ["black", "hispanic", "white", "asian"]
labels = {"black": "Black", "hispanic": "Hispanic", "white": "White", "asian": "Asian"}

totals = panel.groupby(["state", "fall_year"])[groups + ["ftft"]].sum().reset_index()
fig, axes = plt.subplots(2, 4, figsize=(16, 7), sharex=True)
for j, g in enumerate(groups):
    for st, ls in [("NY", "-"), ("PA", "--")]:
        d = totals[totals["state"] == st]
        axes[0, j].plot(d["fall_year"], d[g], ls, marker="o", label=st, color="black")
        axes[1, j].plot(d["fall_year"], d[g] / d["ftft"] * 100, ls, marker="o", label=st, color="black")
    axes[0, j].set_title(labels[g])
    for i in range(2):
        axes[i, j].axvline(2016.5, color="gray", ls=":", lw=0.8)
axes[0, 0].set_ylabel("First-time students (count)")
axes[1, 0].set_ylabel("Share of first-time students (%)")
axes[0, 0].legend()
plt.tight_layout()
plt.savefig("fig1_trends_by_race.png", dpi=200)
```

## B3. Group-specific parallel trends across measures (*centrality of racism*)

Does Pennsylvania remain a credible counterfactual for each population, on each measure? A failed test here is a finding about the limits of the design, not a null result.

```python
measures = {"count": "{g}", "share": "{g}_share"}
if HAVE_CCD:
    measures["per 1k 12th graders"] = "{g}_per1k"

rows = []
for g in groups:
    for mname, pattern in measures.items():
        r = pretrend_test(pattern.format(g=g))
        rows.append({"group": labels[g], "measure": mname, "F": r["F"], "p": r["p"],
                     "parallel_trends": "holds" if r["p"] >= 0.05 else "FAILS"})
pt_table = pd.DataFrame(rows)
pt_table.pivot(index="group", columns="measure", values="parallel_trends")
```

```python
pt_table  # full detail for your Table 6
```

## B4. Event studies by group, interpreted only where pre-trends hold

Coefficients for groups that fail B3 are reported but flagged, not interpreted causally.

```python
rows = []
for g in groups:
    for mname, pattern in measures.items():
        y = pattern.format(g=g)
        es, tests = event_study(y)
        ok = pt_table.query("group == @labels[@g] and measure == @mname")["parallel_trends"].iloc[0] == "holds"
        rows.append({"group": labels[g], "measure": mname,
                     "2017": es.loc[2017, "coef"], "2018": es.loc[2018, "coef"], "2019": es.loc[2019, "coef"],
                     "pre_F_p": tests["pre_p"], "post_F_p": tests["post_p"],
                     "interpretable": "yes" if ok else "no: pre-trends fail"})
es_by_group = pd.DataFrame(rows)
es_by_group
```

## B5. Institutional context: CUNY vs. SUNY (*voice and insight / historical context*)

New York's public sector is not one system. CUNY enrolls a far larger share of Black and Hispanic students than SUNY. Estimating the effect separately asks whose institutions the average effect describes. Each model compares one NY system to all of Pennsylvania.

```python
pa = panel[panel["ny"] == 0]
systems = {"CUNY": pd.concat([panel[(panel["ny"] == 1) & (panel["cuny"] == 1)], pa]),
           "SUNY / other NY public": pd.concat([panel[(panel["ny"] == 1) & (panel["cuny"] == 0)], pa])}

# who each system enrolls (2016)
print((panel[panel["fall_year"] == 2016].assign(system=lambda d: np.where(d["ny"] == 0, "PA",
        np.where(d["cuny"] == 1, "CUNY", "SUNY/other")))
       .groupby("system")[groups + ["ftft"]].sum()
       .pipe(lambda d: d[groups].div(d["ftft"], axis=0) * 100).round(1)))

rows = []
for name, d in systems.items():
    for y in ["ftft", "pell", "black", "hispanic"]:
        pt = pretrend_test(y, data=d)
        r = did(y, data=d)
        rows.append({"system": name, "outcome": y, "did_coef": r["coef"], "se": r["se"], "p": r["p"],
                     "pretrend_p": pt["p"], "interpretable": "yes" if pt["p"] >= 0.05 else "no"})
pd.DataFrame(rows)
```

## B6. The Pell result under a last-dollar design

Howard reads Pell gains as benefit to lower-income students. Because Excelsior is last-dollar, many Pell recipients receive little or no award. Compare the Pell **count** effect with the Pell **share** effect: if the count rises but the share doesn't, Pell enrollment grew in step with overall enrollment, which is consistent with a broad signal ("college is free") more than with targeted cost relief.

```python
panel["pell_share"] = panel["pell"] / panel["ftft"] * 100
pd.DataFrame([
    {"outcome": "Pell count", **did("pell")},
    {"outcome": "Pell share of FTFT (%)", **did("pell_share")},
    {"outcome": "FTFT count", **did("ftft")},
])
```

## B7. Sensitivity: how sample decisions change the estimate (*numbers are not neutral*)

The same model on alternative samples. If the estimate moves, the sample decision is doing analytic work and belongs in the decision log.

```python
largest = panel.groupby("unitid")["ftft"].mean().nlargest(5).index
samples = {
    "Main sample": panel,
    "Drop combined Penn State": panel[panel["unitid"] != 214777],
    "Drop CUNY": panel[panel["cuny"] == 0],
    "Drop 5 largest institutions": panel[~panel["unitid"].isin(largest)],
}
pd.DataFrame([{"sample": k, "institutions": d["unitid"].nunique(), **did("ftft", data=d)}
              for k, d in samples.items()])
```
