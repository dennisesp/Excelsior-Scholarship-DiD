### Year-by-Year Data Merging

Across each academic year (From 2012-2013 to 2019-2020). I merged three seperate IPEDS data files.

The Three Files Include: 
1. Annual Pell Grant Recipient Counts Per Each Academic Year for first-time, full-time, undergraduate students across schools in NY and PA
2. Race and Gender of first-time, full-time, undergraduate students across schools in NY and PA
3. Tuition and Fees for each year




#### 2012-2013 Data Merge

```python
## Loading in the necessary data frames 
pell_2012_2013=pd.read_csv("Pell Grant/Pell_2012_2013.csv")
race_2012_2013=pd.read_csv("Race/2012_2013_Race.csv")
tuition_2012_2016=pd.read_csv("Tuition and Fees/Tuition_2012-2016.csv")

## Defining the specific columns I want from the tuition data set 
## (I downloaded multiple years within one dataset, so I need to specify the ones for the current year I'm working with.)
columns_to_keep_1213=['unitid', 'IC2015_AY.Published out-of-state tuition and fees 2012-13','IC2015_AY.Published in-state tuition and fees 2012-13', 'IC2015_AY.Published in-district tuition and fees 2012-13']
tuition_2012_2013=tuition_2012_2016[columns_to_keep_1213].copy()
tuition_2012_2013


## Merging the Pell and Race Data Together Based on 'unitid' 
merge_2012_2013_1= pd.merge(pell_2012_2013, race_2012_2013, on=['unitid'], how='inner')

## Merging the merged data set from the previous step with the year's tuition data
merge_2012_2013_final=pd.merge(merge_2012_2013_1, tuition_2012_2013, on=['unitid'], how='inner')

## merge_2012_2013 final, now contains all the data I need for this year. 
## I use this same proccess for the remainder of years 
## (Except for an additional step I had to take with a weird bug in 2014-2015)
```

#### Merging All Years Together 

```python
import re
## listing the data frames 
frames = {
    2012: merge_2012_2013_final,
    2013: merge_2013_2014_final,
    2014: merge_2014_2015_final,
    2015: merge_2015_2016_final,
    2016: merge_2016_2017_final,
    2017: merge_2017_2018_final,
    2018: merge_2018_2019_final,
    2019: merge_2019_2020_final,
}



## cleaning up additional/repeat columns that came from merging (Institution Name, Year, etc.) 

def clean(df, year):
    df = df.copy()
    
    df.columns = [re.sub(r"^[A-Z]+\d{4}[A-Z]?(_RV|_AY)?\.", "", c) for c in df.columns]
    df.columns = [re.sub(r"\s\d{4}-\d{2}(_x|_y)?$", "", c) for c in df.columns]
    
    df = df.drop(columns=[c for c in df.columns if c.startswith(("year", "institution name_y"))], errors="ignore")
    df = df.rename(columns={"institution name_x": "institution name"})
    df = df.loc[:, ~df.columns.duplicated()]
    df["fall_year"] = year
    return df

## Clean Panel with every year's data

panel = pd.concat([clean(df, yr) for yr, df in frames.items()], ignore_index=True)


```

### Merging County Data

Here, I realized I didnt add county data FIPS to the original pull, so I have to merge it into the panel created in the previous step

However, for some reason, everytime I downloaded the "FIPS County Code" variable per institution on IPEDS, I would only get the county name. The County Code will be essential when I merge the Bureau of Labor Statistics Data, so, I decided to match the IPEDS downloaded county names, with a dataset that has both county names and codes. Luckily, a very kind Github user had just the CSV FILE with the countynames and codes I needed. 

Below you'll see me load in both the IPEDS fips county name, as well as the external county name with county code document

```python
county_data=pd.read_csv("Other Data/county_fips_master.csv", encoding='cp1252')
fips_county=pd.read_csv("Other Data/2019_county.csv")
```

```python
## Selecting only PA and NY county codes

county_data = county_data[county_data["state_abbr"].isin(["NY", "PA"])].copy()


## Cleaning the county names for merging. 

county_data["county"] = (
    county_data["county_name"]
    .str.replace(r"[^A-Za-z0-9]", "", regex=True)
    .str.lower()
)


## Renaming "state" with "state_abbr" for easier merging/so column_name is the same as the one in the IPEDS county df
county_data["state"] = county_data["state_abbr"]

## Making sure county_data is a string, and keeping leading 0s
county_data["fips"] = county_data["fips"].astype(str).str.zfill(5)   

county_data = county_data[["state", "county", "fips"]]
```

```python
## creating a Fips "state" columm by extracting 
fips_county["state"] = fips_county["HD2019.Fips County code"].str.extract(r",\s*(NY|PA)$")



fips_county["county"] = (
    fips_county["HD2019.Fips County code"]
    .str.replace(r",\s*(NY|PA)$", "", regex=True)
    .str.replace(r"[^A-Za-z0-9]", "", regex=True)
    .str.lower()
)



fips_county = fips_county.merge(county_data, on=["state", "county"], how="left")
print(fips_county["fips"].isna().sum(), "institutions with no FIPS match")
```

```python
fips_county
```

```python
## ensuring I get one row per institution

fips_county = fips_county.drop_duplicates("unitid")  

n_before = len(panel)

panel = panel.merge(
    fips_county[["unitid", "state", "county", "fips"]],
    on="unitid",
    how="left",
    validate="many_to_one"   
)

print(n_before, "->", len(panel), "rows")                 # these should match
print(panel["fips"].isna().sum(), "rows missing FIPS")
```

## Adding BLS Data

Now, I'm adding each year's lagged unemployment per county BLS data (2011-2018)

```python
BLS2011=pd.read_excel("BLS Data/laucnty11.xlsx")
BLS2012=pd.read_excel("BLS Data/laucnty12.xlsx")
BLS2013=pd.read_excel("BLS Data/laucnty13.xlsx")
BLS2014=pd.read_excel("BLS Data/laucnty14.xlsx")
BLS2015=pd.read_excel("BLS Data/laucnty15.xlsx")
BLS2016=pd.read_excel("BLS Data/laucnty16.xlsx")
BLS2017=pd.read_excel("BLS Data/laucnty17.xlsx")
BLS2018=pd.read_excel("BLS Data/laucnty18.xlsx")
```

```python
bls_list = [BLS2011, BLS2012, BLS2013, BLS2014, BLS2015, BLS2016, BLS2017, BLS2018]

frames = []
for d in bls_list:
    d = d.iloc[:, [1, 2, 4, 8]].copy()          # state FIPS, county FIPS, year, unemployment rate
    d.columns = ["state_fips", "county_fips", "year", "unemp"]
    frames.append(d)

bls = pd.concat(frames, ignore_index=True)

# convert to numbers; title, header, and footer rows become NaN and get dropped
for c in bls.columns:
    bls[c] = pd.to_numeric(bls[c], errors="coerce")
bls = bls.dropna()

bls = bls[bls["state_fips"].isin([36, 42])]      # NY and PA

# rebuild 5-digit FIPS with leading zeros, e.g. 36 + 1 -> "36001"
bls["fips"] = (bls["state_fips"].astype(int).astype(str).str.zfill(2)
               + bls["county_fips"].astype(int).astype(str).str.zfill(3))
bls["year"] = bls["year"].astype(int)
bls = bls[["fips", "year", "unemp"]]

print(bls.groupby("year").size())   # should be 129 for every year
```

```python
panel["fips"] = panel["fips"].astype(str).str.zfill(5)   # make sure the panel's FIPS is a 5-digit string too
panel["unemp_year"] = panel["fall_year"] - 1

panel = panel.merge(bls, left_on=["fips", "unemp_year"], right_on=["fips", "year"],
                    how="left").rename(columns={"unemp": "unemp_lag"}).drop(columns="year")

print(panel["unemp_lag"].isna().sum(), "rows missing unemployment")
```

```python
control_2012=pd.read_csv("Other Data/2012_control.csv")
control_2019=pd.read_csv("Other Data/2019_control.csv")
```

```python
# rename the control columns
control_2012 = control_2012.rename(columns={"IC2012_RV.Institutional control or affiliation": "control_2012"})
control_2019 = control_2019.rename(columns={"HD2019.Control of institution": "control"})

# flag public institutions
control_2012["public_2012"] = control_2012["control_2012"].str.startswith("Public")
control_2019["public"] = control_2019["control"].str.startswith("Public")

# check whether any institution switched between public and private
control_2019 = control_2019.merge(control_2012[["unitid", "public_2012"]], on="unitid", how="left")
print((control_2019["public"] != control_2019["public_2012"]).sum(), "institutions switched")



# merge into the panel
panel = panel.merge(control_2019[["unitid", "control", "public"]], on="unitid", how="left")

# keep public institutions, drop Penn State World Campus (online)
panel = panel[(panel["public"] == True) & (panel["unitid"] != 479956)]

print(panel.groupby(["fall_year", "state"])["unitid"].nunique().unstack())
```

```python
psu_main_id = 214777   # Pennsylvania State University-Main Campus

# check first: does Main Campus jump in 2019?
print(panel.loc[panel["unitid"] == psu_main_id, ["fall_year", "Grand total"]])



psu = panel["institution name"].str.startswith("Pennsylvania State University", na=False)


count_cols = [c for c in panel.columns if c.endswith(("total", "men", "women"))] + [
    "Number of full-time first-time undergraduates awarded Pell grants"]

psu_rows = panel[psu].groupby("fall_year", as_index=False)[count_cols].sum()
main_info = panel[panel["unitid"] == psu_main_id].drop(columns=count_cols)
psu_rows = psu_rows.merge(main_info, on="fall_year", how="left")

panel = pd.concat([panel[~psu], psu_rows], ignore_index=True)
panel["Percent of full-time first-time undergraduates awarded Pell grants"] = (
    panel["Number of full-time first-time undergraduates awarded Pell grants"] / panel["Grand total"] * 100)

print(panel.groupby(["fall_year", "state"])["unitid"].nunique().unstack())   # PA should be 38 every year
```
